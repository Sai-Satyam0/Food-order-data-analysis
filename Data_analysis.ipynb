## 4. Missing values
Find which columns actually have nulls, and how many.

```python
null_value_columns = []
for i in df.columns:
    if df[i].isnull().sum() > 0:
        null_value_columns.append(i)
    else:
        continue

null_value_columns
```

```python
df[null_value_columns].isnull().sum()
```

As we can see the cancellation_reason column has 49132 null values (NO Reason) out of 50,000. which simply means the 49132 orders aren't Cancelled they have deliverd successfully. 
Also not to mention there are 868 null values in late_delivery,Customer_Rating and Actual_delivery_time_minutes columns, which indicates these value comes from the delivery that got cancelled because 49132 + 868 = 50,000 which is the actual length of the dataset. Hence there are no orders with tip amount, late delivery, customer rating. actual delivery timme minutes that has delivered.



## 5. Cleaning

```python
df["cancellation_reason"] = df["cancellation_reason"].fillna("Not Cancelled")
df["cancellation_reason"].value_counts()
```

```python
# A cancelled order was never tipped
df["tip_amount"] = df["tip_amount"].fillna(0)
```

```python
# No delivery happened, so there's no real delivery time or lateness flag to impute.
# Filling with the average of completed orders keeps the column numeric for Power BI
# without pretending the order was on time.
df["actual_delivery_time_minutes"] = df["actual_delivery_time_minutes"].fillna(df["actual_delivery_time_minutes"].mean())
df["late_delivery"] = df["late_delivery"].fillna(df["late_delivery"].mean())
```

```python
# No customer rating was ever given for a cancelled order
df["customer_rating"] = df["customer_rating"].fillna(df["customer_rating"].mean())
```

### Confirming nothing is missing anymore

```python
df.isnull().sum().sum()
```

## 6. Fix data types
`order_date` and `order_timestamp` are read in as text, converting them to real dates makes them usable for time-based visuals in Power BI

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["order_timestamp"] = pd.to_datetime(df["order_timestamp"])
df.dtypes[["order_date", "order_timestamp"]]
```
